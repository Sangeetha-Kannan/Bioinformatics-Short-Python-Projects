Import libraries

```python
from Bio import pairwise2
from Bio.pairwise2 import format_alignment
from Bio import SeqIO
```

### Step 2: Define Functions for Sequence Alignment

We'll use Biopython's **pairwise2** module to perform pairwise sequence alignment

```python
# Function to perform pairwise sequence alignment
def align_sequences(seq1, seq2):
    alignments = pairwise2.align.globalxx(seq1, seq2)
    return alignments
```

```python
# Function to print the best alignment
def print_best_alignment(alignments):
    best_alignment = alignments[0]
    print(format_alignment(*best_alignment))
```
